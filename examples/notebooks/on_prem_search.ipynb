# Search your on-prem CHEESE installation

Select **CHEESE (on-prem)**. The kernel supplies your company's API address. You can also run this notebook in an environment with the deployment SDK and `DEEPMEDCHEM_ON_PREM_URL` set to your internal origin, for example `https://cheese.your-pharma-company.com`.

```python
from deepmedchem import Client

with Client(on_prem=True) as cheese:
    libraries = [library["database_id"] for library in cheese.catalog()["libraries"]]
print(libraries)
```

The pilot includes Mcule AA In-Stock. Select an installed library from the list above to search a different catalogue.

```python
database = "mcule-aa-in-stock"
with Client(on_prem=True) as cheese:
    result = cheese.search("O=C(O)CNc1ccc(Br)c(Cl)c1C(=O)O", database=database, limit=3)
for hit in result.hits:
    print(hit.catalog_id, hit.score)
```
