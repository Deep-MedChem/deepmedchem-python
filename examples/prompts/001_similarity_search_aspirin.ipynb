Find the 5 most similar molecules to aspirin in the Enamine REAL database and show their SMILES, similarity score, and price.

```python
from deepmedchem import Client

DATABASE = "enamine-real-v5a"
ASPIRIN_SMILES = "CC(=O)Oc1ccccc1C(=O)O"

with Client() as dmc:
    result = dmc.search(ASPIRIN_SMILES, database=DATABASE, method="shape", limit=5)

for hit in result.hits:
    price = f"${hit.price}" if hit.price is not None else "unavailable"
    print(f"{hit.rank}  score={hit.score:.4f}  price={price}  {hit.smiles}")
```
