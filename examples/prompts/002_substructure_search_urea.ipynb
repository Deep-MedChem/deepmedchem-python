## Prompt

> Search the Freedom Space 5 database for molecules containing an acyclic urea substructure and report how many matches were found and their SMILES.

```python
import deepmedchem as dmc

DATABASE = "freedom-space-5"
ACYCLIC_UREA_SMARTS = "[N;R0]C(=O)[N;R0]"

result = dmc.substructure(
    ACYCLIC_UREA_SMARTS,
    format="smarts",
    database=DATABASE,
    limit=5,
)

print(f"acyclic urea: {len(result)} hits")
for hit in result.hits:
    print(f"  {hit.rank}  {hit.smiles}")
```
